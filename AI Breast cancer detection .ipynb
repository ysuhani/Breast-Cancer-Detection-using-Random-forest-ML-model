```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn import metrics
import numpy as np
```

```python
#Import scikit-learn dataset library
from sklearn import datasets

#Load dataset
cancer = datasets.load_breast_cancer()
```

```python
# print the names of the 13 features
print("Features: ", cancer.feature_names)

# print the label type of cancer('malignant' 'benign')
print("Labels: ", cancer.target_names)
```

```python
# Split dataset into training set and test set (30% test, 70% train)
X_train, X_test, y_train, y_test = train_test_split(cancer.data, cancer.target, test_size=0.3, random_state=109)
```

```python
# Create and train the Random Forest model
clf = RandomForestClassifier(n_estimators=100, random_state=42)
clf.fit(X_train, y_train)
```

```python
# Predict the response for the test dataset
y_pred = clf.predict(X_test)
```

```python
# Model Accuracy, Precision, and Recall on Test Data
print("Accuracy on Test Data:", metrics.accuracy_score(y_test, y_pred))
print("Precision on Test Data:", metrics.precision_score(y_test, y_pred))
print("Recall on Test Data:", metrics.recall_score(y_test, y_pred))
```

```python
# Use K-Fold Cross-Validation (e.g., 5-fold)
from sklearn.model_selection import KFold
kf = KFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = cross_val_score(clf, X_train, y_train, cv=kf)

# Report the performance from K-fold cross-validation
print("\nK-Fold Cross-Validation Results:")
print("Cross-Validation Scores:", cv_scores)
print("Mean Cross-Validation Score:", np.mean(cv_scores))
```

```python
# Hyperparameter tuning using GridSearchCV
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'bootstrap': [True, False]
}

# Use GridSearchCV for hyperparameter tuning
grid_search = GridSearchCV(estimator=clf, param_grid=param_grid, cv=3, n_jobs=-1, verbose=2)
grid_search.fit(X_train, y_train)

# Get the best parameters from the grid search
best_params = grid_search.best_params_
print("\nBest Hyperparameters from GridSearchCV:", best_params)
```

```python
# Step 5: Evaluate the model with best hyperparameters on the test set
best_model = grid_search.best_estimator_
y_pred_best = best_model.predict(X_test)

# Report the performance after tuning hyperparameters
print("\nFinal Performance after Hyperparameter Tuning:")
print("Accuracy on Test Data:", metrics.accuracy_score(y_test, y_pred_best))
print("Precision on Test Data:", metrics.precision_score(y_test, y_pred_best))
print("Recall on Test Data:", metrics.recall_score(y_test, y_pred_best))
print("f1-score on Test Data:", metrics.f1_score(y_test, y_pred_best))

# Optionally, print a detailed Classification Report
print("\nClassification Report:\n", metrics.classification_report(y_test, y_pred_best))
```
